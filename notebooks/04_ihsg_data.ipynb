```python
#Import Libraries
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
#Download IHSG Data
ihsg = yf.download("^JKSE", start="2024-10-01", end="2026-06-01")
print(f"IHSG data shape: {ihsg.shape}")
print(ihsg.tail(5))
```

```python
#Clean & Save IHSG Data

# Reset index
ihsg = ihsg.reset_index()
ihsg.columns = ["date", "close", "high", "low", "open", "volume"]
ihsg["date"] = pd.to_datetime(ihsg["date"])

# Keep only date and close price
ihsg = ihsg[["date", "close"]]

print(f"Date range: {ihsg['date'].min()} -> {ihsg['date'].max()}")
print(f"Close price range: {ihsg['close'].min():.2f} -> {ihsg['close'].max():.2f}")
print(ihsg.head(5))

# Save
ihsg.to_csv("../data/ihsg/ihsg_data.csv", index=False)
```

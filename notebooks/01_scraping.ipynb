```python
# multi-page scraping with date filter
def parse_page(driver):
    soup = BeautifulSoup(driver.page_source, "html.parser")
    articles = soup.find_all("div", class_="articleItem")
    
    data = []
    for article in articles:
        title = article.find("a")
        title_text = title.text.strip() if title else None
        url = title["href"] if title else None
        date = article.find("div", class_="articlePost-date")
        date_text = date.text.strip() if date else None
        category = article.find("div", class_="articlePost-subtitle")
        category_text = category.text.strip() if category else None
        
        data.append({
            "title": title_text,
            "url": url,
            "date": date_text,
            "category": category_text
        })
    return data

# Scrape page by page
all_data = []
page = 1
stop = False

while not stop:
    print(f"Scraping page {page}...")
    
    # Parse current page
    page_data = parse_page(driver)
    
    for item in page_data:
        all_data.append(item)
        # Stop if article older than October 2024
        if item["date"] and "2024" in item["date"]:
            month = item["date"].split(" ")[1] if item["date"] else ""
            months_before = ["Januari", "Februari", "Maret", "April", "Mei", "Juni", "Juli", "Agustus", "September"]
            if month in months_before:
                print(f"Reached pre-Prabowo article: {item['date']} — stopping.")
                stop = True
                break
    
    if not stop:
        # Go to next page
        try:
            next_btn = driver.find_element(By.LINK_TEXT, "Next")
            next_btn.click()
            time.sleep(3)
            page += 1
        except:
            print("No more pages.")
            break

df = pd.DataFrame(all_data)
print(f"\nTotal articles scraped: {len(df)}")
print(df.head())
```

```python
# full multi-page scraper
def parse_page(driver):
    soup = BeautifulSoup(driver.page_source, "html.parser")
    articles = soup.find_all("div", class_="articleItem")
    
    data = []
    for article in articles:
        title = article.find("a")
        title_text = title.text.strip() if title else None
        url = title["href"] if title else None
        date = article.find("div", class_="articlePost-date")
        date_text = date.text.strip() if date else None
        category = article.find("div", class_="articlePost-subtitle")
        category_text = category.text.strip() if category else None
        
        data.append({
            "title": title_text,
            "url": url,
            "date": date_text,
            "category": category_text
        })
    return data

def is_before_october_2024(date_text):
    """Check if article is before Prabowo's inauguration (Oct 2024)"""
    if not date_text:
        return False
    months_map = {
        "Januari": 1, "Februari": 2, "Maret": 3, "April": 4,
        "Mei": 5, "Juni": 6, "Juli": 7, "Agustus": 8,
        "September": 9, "Oktober": 10, "November": 11, "Desember": 12
    }
    parts = date_text.strip().split(" ")
    if len(parts) == 3:
        day, month, year = parts
        year = int(year)
        month_num = months_map.get(month, 0)
        if year < 2024 or (year == 2024 and month_num < 10):
            return True
    return False

#Start scraping from page 1
driver.get("https://www.kompas.com/tag/prabowo")
time.sleep(3)

all_data = []
page = 1
stop = False

while not stop:
    print(f"Scraping page {page}...")
    page_data = parse_page(driver)
    
    for item in page_data:
        if is_before_october_2024(item["date"]):
            print(f"Reached pre-inauguration article: {item['date']} — stopping.")
            stop = True
            break
        all_data.append(item)
    
    if not stop:
        try:
            next_btn = driver.find_element(By.LINK_TEXT, "Next")
            driver.execute_script("arguments[0].click();", next_btn)
            time.sleep(3)
            page += 1
        except:
            print("No more pages.")
            break

df = pd.DataFrame(all_data)
print(f"\nTotal articles scraped: {len(df)}")
print(df.head(10))
```

```python
#Extract Articles with Date from Tempo
soup = BeautifulSoup(driver.page_source, "html.parser")

container = soup.find("div", class_="flex flex-col divide-y divide-neutral-500")
articles = container.find_all("a", href=True) if container else []

# Filter only article links (skip duplicates and empty text)
seen = set()
data = []
for a in articles:
    href = a.get("href", "")
    title = a.text.strip()
    
    # Skip empty titles and duplicates
    if not title or href in seen:
        continue
    seen.add(href)
    
    # Get date — look for time tag near the article
    parent = a.find_parent()
    time_tag = parent.find("time") if parent else None
    date_text = time_tag.get("datetime") or time_tag.text.strip() if time_tag else None
    
    data.append({
        "title": title,
        "url": "https://www.tempo.co" + href,
        "date": date_text
    })

import pandas as pd
df_tempo = pd.DataFrame(data)
print(f"Articles found: {len(df_tempo)}")
print(df_tempo.head(10))
```
